```python

from queue import Queue

def record(trace, node, queue_items, visited):
    if trace is not None:
        trace.append((node, queue_items, visited))

def bfs_tree(tree, root, goal, trace=None):
    queue = Queue()
    queue.put([root])
    visited = []

    while not queue.empty():
        path = queue.get()
        node = path[-1]
        visited.append(node)

        if node == goal:
            record(trace, node, [p[-1] for p in queue.queue], visited)
            return path

        for child in tree[node]:
            queue.put(path + [child])

        record(trace, node, [p[-1] for p in queue.queue], visited)

    return None

def bfs_graph(graph, root, goal, trace=None):
    queue = [[root]]
    visited = set()

    while len(queue) > 0:
        path = queue.pop(0)
        node = path[-1]
        
        if node in visited:
            continue
            
        visited.add(node)

        if node == goal:
            record(trace, node, [p[-1] for p in queue], list(visited))
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append(path + [neighbor])

        record(trace, node, [p[-1] for p in queue], list(visited))

    return None
```
